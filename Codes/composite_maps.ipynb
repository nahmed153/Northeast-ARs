GEOPOTENTIAL HEIGHTS

```
!pip install basemap
from mpl_toolkits.basemap import Basemap
import matplotlib.pyplot as plt
import numpy as np
from netCDF4 import Dataset # Import Dataset here

data1 = Dataset(r'/content/drive/MyDrive/Northeast_ARs/850mb_mean.nc')

lats_full = data1.variables['lat'][:]
lons_original_full = data1.variables['lon'][:]
hgt_full = data1.variables['hgt'][:] # Ensure this is 'hgt' for 850mb_anomaly.nc

# Adjust longitudes to be -180 to 180 for better compatibility with Basemap
lons_full_adjusted = np.where(lons_original_full > 180, lons_original_full - 360, lons_original_full)

# Define the desired geographical boundaries
lat_min = 27.5
lat_max = 65.0
lon_min = -130.0 # 130 degrees West
lon_max = -60.0  # 60 degrees West

# Filter latitudes and longitudes
lat_indices = np.where((lats_full >= lat_min) & (lats_full <= lat_max))
lons_indices = np.where((lons_full_adjusted >= lon_min) & (lons_full_adjusted <= lon_max))

lats = lats_full[lat_indices]
lons = lons_full_adjusted[lons_indices]

# Assuming hgt is (time, lat, lon)
hgt = hgt_full[:, lat_indices[0][:, np.newaxis], lons_indices[0]]

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111)

# Define Basemap with the new, reduced geographical boundaries
mp = Basemap(projection='merc',
             llcrnrlon=lon_min,
             llcrnrlat=lat_min,
             urcrnrlon=lon_max,
             urcrnrlat=lat_max,
             resolution='i')

# Create meshgrid for lat/lon with the filtered and adjusted longitudes
lon_grid, lat_grid = np.meshgrid(lons, lats)

# Transform lat/lon to map coordinates
x, y = mp(lon_grid, lat_grid)

im = mp.pcolormesh(x, y, hgt[0,:,:], cmap='viridis', vmin=hgt.min(), vmax=hgt.max(), edgecolors='none')

# Add contour lines for geopotential height
# Define contour levels at intervals of 20
contour_levels = np.arange(np.floor(hgt.min()/20)*20, np.ceil(hgt.max()/20)*20 + 1, 20)
cs = mp.contour(x, y, hgt[0,:,:], levels=contour_levels, colors='k', linewidths=0.8)
plt.clabel(cs, inline=1, fontsize=10, fmt='%1.0f') # Add labels to the contour lines, fontsize 10

# Add coastlines
mp.drawcoastlines()
mp.drawcountries()
# mp.drawstates() # Uncomment if states are desired


# Add a colorbar
cbar = plt.colorbar(im, shrink=0.7)
cbar.set_label('Geopotential Height (m)')

# Add a title

plt.show()
```

```
# Define the paths to the netCDF files for u and v wind components for 850 hPa MEAN
file_path_uwnd = r'/content/drive/MyDrive/Northeast_ARs/850mbwind_mean_u.nc'
file_path_vwnd = r'/content/drive/MyDrive/Northeast_ARs/850mbwind_mean_v.nc'

# Load the datasets
try:
    data_uwnd = Dataset(file_path_uwnd)
    data_vwnd = Dataset(file_path_vwnd)
except FileNotFoundError as e:
    print(f"Error: File not found. Please check paths. {e}")
    exit()

# Extract full data arrays for lat, lon, uwnd, and vwnd
lats_full = data_uwnd.variables['lat'][:] # Latitudes should be the same for both
lons_original_full = data_uwnd.variables['lon'][:] # Longitudes should be the same
uwnd_full = data_uwnd.variables['uwnd'][:] # U-component of wind
vwnd_full = data_vwnd.variables['vwnd'][:] # V-component of wind

# Adjust longitudes to be -180 to 180 for better compatibility with Basemap
lons_full_adjusted = np.where(lons_original_full > 180, lons_original_full - 360, lons_original_full)

# Define the desired geographical boundaries (from previous cells)
lat_min = 27.5
lat_max = 65.0
lon_min = -130.0 # 130 degrees West
lon_max = -60.0  # 60 degrees West

# Filter latitudes and longitudes
lat_indices = np.where((lats_full >= lat_min) & (lats_full <= lat_max))
lons_indices = np.where((lons_full_adjusted >= lon_min) & (lons_full_adjusted <= lon_max))

lats_filtered = lats_full[lat_indices]
lons_filtered = lons_full_adjusted[lons_indices]

# Filter uwnd and vwnd data to match the new lat/lon dimensions
# Assuming uwnd/vwnd are (time, lat, lon) and we're taking the first time step
uwnd_sliced = uwnd_full[0, lat_indices[0][:, np.newaxis], lons_indices[0]]
vwnd_sliced = vwnd_full[0, lat_indices[0][:, np.newaxis], lons_indices[0]]

# Calculate wind speed magnitude for the colored plot
wind_speed = np.sqrt(uwnd_sliced**2 + vwnd_sliced**2)

# Print min/max values for debugging
print(f"uwnd_sliced min: {uwnd_sliced.min():.2f}, max: {uwnd_sliced.max():.2f}")
print(f"vwnd_sliced min: {vwnd_sliced.min():.2f}, max: {vwnd_sliced.max():.2f}")
print(f"wind_speed min: {wind_speed.min():.2f}, max: {wind_speed.max():.2f}")

fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111)

# Define Basemap with the new, reduced geographical boundaries
mp = Basemap(projection='merc',
             llcrnrlon=lon_min,
             llcrnrlat=lat_min,
             urcrnrlon=lon_max,
             urcrnrlat=lat_max,
             resolution='i')

# Create meshgrid for lat/lon with the filtered and adjusted longitudes
lon_grid, lat_grid = np.meshgrid(lons_filtered, lats_filtered)

# Transform lat/lon to map coordinates
x, y = mp(lon_grid, lat_grid)

# Plot the wind speed magnitude as a colored background
im = mp.pcolormesh(x, y, wind_speed, cmap='RdBu_r', vmin=0, vmax=wind_speed.max(), edgecolors='none')

# Add a colorbar for wind speed
cbar = plt.colorbar(im, shrink=0.7)
cbar.set_label('850 hPa Wind Speed Mean (m/s)') # Updated label to 'Mean'


# Subsample the data for plotting wind vectors to avoid clutter
skip = 3 # Plot every Nth vector (even more density)
x_sub = x[::skip, ::skip]
y_sub = y[::skip, ::skip]
uwnd_sub = uwnd_sliced[::skip, ::skip]
vwnd_sub = vwnd_sliced[::skip, ::skip]

# Normalize wind vectors to have constant length
wind_magnitude_sub = np.sqrt(uwnd_sub**2 + vwnd_sub**2)
# Avoid division by zero for areas with no wind (magnitude is 0)
u_norm = np.divide(uwnd_sub, wind_magnitude_sub, out=np.zeros_like(uwnd_sub), where=wind_magnitude_sub!=0)
v_norm = np.divide(vwnd_sub, wind_magnitude_sub, out=np.zeros_like(vwnd_sub), where=wind_magnitude_sub!=0)

# Plot wind vectors on top of the map, all with uniform length
# Adjust 'scale' and 'width' for desired arrow appearance
Q = mp.quiver(x_sub, y_sub, u_norm, v_norm,
              units='width', scale=25, width=0.004, color='k', headlength=5, headwidth=4, headaxislength=4)

# Add coastlines and countries
mp.drawcoastlines()
mp.drawcountries()

# Add a title
plt.show()
```

```
# Define the paths to the netCDF files for u and v wind components for 1000 hPa MEAN
file_path_uwnd = r'/content/drive/MyDrive/Northeast_ARs/1000mbwind_mean_u.nc'
file_path_vwnd = r'/content/drive/MyDrive/Northeast_ARs/1000mbwind_mean_v.nc'

# Load the datasets
try:
    data_uwnd = Dataset(file_path_uwnd)
    data_vwnd = Dataset(file_path_vwnd)
except FileNotFoundError as e:
    print(f"Error: File not found. Please check paths. {e}")
    exit()

# Extract full data arrays for lat, lon, uwnd, and vwnd
lats_full = data_uwnd.variables['lat'][:] # Latitudes should be the same for both
lons_original_full = data_uwnd.variables['lon'][:] # Longitudes should be the same
uwnd_full = data_uwnd.variables['uwnd'][:] # U-component of wind
vwnd_full = data_vwnd.variables['vwnd'][:] # V-component of wind

# Adjust longitudes to be -180 to 180 for better compatibility with Basemap
lons_full_adjusted = np.where(lons_original_full > 180, lons_original_full - 360, lons_original_full)

# Define the desired geographical boundaries (from previous cells)
lat_min = 27.5
lat_max = 65.0
lon_min = -130.0 # 130 degrees West
lon_max = -60.0  # 60 degrees West

# Filter latitudes and longitudes
lat_indices = np.where((lats_full >= lat_min) & (lats_full <= lat_max))
lons_indices = np.where((lons_full_adjusted >= lon_min) & (lons_full_adjusted <= lon_max))

lats_filtered = lats_full[lat_indices]
lons_filtered = lons_full_adjusted[lons_indices]

# Filter uwnd and vwnd data to match the new lat/lon dimensions
# Assuming uwnd/vwnd are (time, lat, lon) and we're taking the first time step
uwnd_sliced = uwnd_full[0, lat_indices[0][:, np.newaxis], lons_indices[0]]
vwnd_sliced = vwnd_full[0, lat_indices[0][:, np.newaxis], lons_indices[0]]

# Calculate wind speed magnitude for the colored plot
wind_speed = np.sqrt(uwnd_sliced**2 + vwnd_sliced**2)

# Print min/max values for debugging
print(f"uwnd_sliced min: {uwnd_sliced.min():.2f}, max: {uwnd_sliced.max():.2f}")
print(f"vwnd_sliced min: {vwnd_sliced.min():.2f}, max: {vwnd_sliced.max():.2f}")
print(f"wind_speed min: {wind_speed.min():.2f}, max: {wind_speed.max():.2f}")

fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111)

# Define Basemap with the new, reduced geographical boundaries
mp = Basemap(projection='merc',
             llcrnrlon=lon_min,
             llcrnrlat=lat_min,
             urcrnrlon=lon_max,
             urcrnrlat=lat_max,
             resolution='i')

# Create meshgrid for lat/lon with the filtered and adjusted longitudes
lon_grid, lat_grid = np.meshgrid(lons_filtered, lats_filtered)

# Transform lat/lon to map coordinates
x, y = mp(lon_grid, lat_grid)

# Plot the wind speed magnitude as a colored background with consistent vmax=8.71
im = mp.pcolormesh(x, y, wind_speed, cmap='RdBu_r', vmin=0, vmax=8.71, edgecolors='none')

# Add a colorbar for wind speed
cbar = plt.colorbar(im, shrink=0.7)
cbar.set_label('1000 hPa Wind Speed Mean (m/s)') # Updated label

# Subsample the data for plotting wind vectors to avoid clutter
skip = 3 # Plot every Nth vector (even more density)
x_sub = x[::skip, ::skip]
y_sub = y[::skip, ::skip]
uwnd_sub = uwnd_sliced[::skip, ::skip]
vwnd_sub = vwnd_sliced[::skip, ::skip]

# Normalize wind vectors to have constant length
wind_magnitude_sub = np.sqrt(uwnd_sub**2 + vwnd_sub**2)
# Avoid division by zero for areas with no wind (magnitude is 0)
u_norm = np.divide(uwnd_sub, wind_magnitude_sub, out=np.zeros_like(uwnd_sub), where=wind_magnitude_sub!=0)
v_norm = np.divide(vwnd_sub, wind_magnitude_sub, out=np.zeros_like(vwnd_sub), where=wind_magnitude_sub!=0)

# Plot wind vectors on top of the map, all with uniform length
# Adjust 'scale' and 'width' for desired arrow appearance
Q = mp.quiver(x_sub, y_sub, u_norm, v_norm,
              units='width', scale=25, width=0.004, color='k', headlength=5, headwidth=4, headaxislength=4)

# Add coastlines and countries
mp.drawcoastlines()
mp.drawcountries()

# Add a title
plt.show()
```

```
file_path = r'/content/drive/MyDrive/Northeast_ARs/storm_type_plots/NW_prate_mean.nc'

# Define the desired geographical boundaries
lat_min = 27.5
lat_max = 65.0
lon_min = -130.0 # 130 degrees West
lon_max = -60.0  # 60 degrees West

try:
    data1 = Dataset(file_path)
except FileNotFoundError:
    print(f"Error: File not found at {file_path}. Please check the path.")
    exit()

lats_full = data1.variables['lat'][:]
lons_original_full = data1.variables['lon'][:]
prate_full = data1.variables['prate'][:]

# Adjust longitudes to be -180 to 180 for better compatibility with Basemap
lons_full_adjusted = np.where(lons_original_full > 180, lons_original_full - 360, lons_original_full)

# Filter latitudes and longitudes based on defined boundaries
lat_indices = np.where((lats_full >= lat_min) & (lats_full <= lat_max))
lons_indices = np.where((lons_full_adjusted >= lon_min) & (lons_full_adjusted <= lon_max))

lats_filtered = lats_full[lat_indices]
lons_filtered = lons_full_adjusted[lons_indices]

# Filter prate data to match the new lat/lon dimensions
# Check the number of dimensions for 'prate_full'
if prate_full.ndim == 3: # Assuming (time, lat, lon)
    prate_sliced = prate_full[0, lat_indices[0][:, np.newaxis], lons_indices[0]]
elif prate_full.ndim == 2: # Assuming (lat, lon)
    prate_sliced = prate_full[lat_indices[0][:, np.newaxis], lons_indices[0]]
else:
    print("Unexpected dimensions for 'prate' variable. Expected 2 or 3 dimensions.")
    exit()

print(f"Min Precipitation Rate: {prate_sliced.min():.2f}, Max Precipitation Rate: {prate_sliced.max():.2f}")

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111)

# Define Basemap for the plot
mp = Basemap(projection='merc',
             llcrnrlon=lon_min,
             llcrnrlat=lat_min,
             urcrnrlon=lon_max,
             urcrnrlat=lat_max,
             resolution='i',
             ax=ax)

# Create meshgrid for lat/lon with the filtered and adjusted longitudes
lon_grid, lat_grid = np.meshgrid(lons_filtered, lats_filtered)

# Transform lat/lon to map coordinates
x, y = mp(lon_grid, lat_grid)

# Plot the prate data as a colored background
im = mp.pcolormesh(x, y, prate_sliced, cmap='YlGn',
                   vmin=prate_sliced.min(), vmax=prate_sliced.max(),
                   edgecolors='none', ax=ax)

# Add contour lines for precipitation rate
# Adjust contour levels based on the data's range
contour_interval = 1.0 # Can be adjusted based on the data's range
contour_levels = np.arange(np.floor(prate_sliced.min()/contour_interval)*contour_interval,
                           np.ceil(prate_sliced.max()/contour_interval)*contour_interval + contour_interval,
                           contour_interval)

cs = mp.contour(x, y, prate_sliced, levels=contour_levels, colors='k', linewidths=0.8, ax=ax)
ax.clabel(cs, inline=1, fontsize=14, fmt='%1.1f') # Format to one decimal place for precipitation values, increased fontsize

# Add coastlines and countries
mp.drawcoastlines(ax=ax)
mp.drawcountries(ax=ax)

# Add a colorbar
cbar = fig.colorbar(im, ax=ax, shrink=0.7)
cbar.set_label('Precipitation Rate Mean (mm/day)', fontsize=12) # Increased fontsize for colorbar label

# Add a title
ax.set_title('Precipitation Rate Mean', fontsize=12)

# Adjust layout to prevent labels from being cut off
plt.tight_layout()

plt.show()
plt.close(fig)
```
